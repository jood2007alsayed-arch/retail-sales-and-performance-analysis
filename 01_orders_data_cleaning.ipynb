```python
df = pd.read_csv("a1_orders_messy.csv", dtype={"Order ID": "string"}, na_values=["", "NA", "-", "n/a"])
df.head()
df.info()
df.describe(include="all")
```

Q1:
befor the last step we had:

Columns with missing values: Total (SAR), Order Date, and Rating

Wrong data types: Order ID should be a string, Total (SAR) should be numeric, and Order Date should be a datetime

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)


df = df.rename(columns={"total_sar": "total"})

print(df.columns)
```

Q2:
Yes, all column names are now clean and predictable

```python
df["total"] = (
    df["total"]
    .astype(str)
    .str.replace(r"[^0-9.-]", "", regex=True)
)

df["total"] = pd.to_numeric(df["total"], errors="coerce")


df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")


print(df.dtypes)
```

```python
df["total"].isna().sum()
```

Q3:
 4 values became NaN because they contained non-numeric or missing values, so they could not be converted to numbers

```python
print(df["city"].value_counts())

df["customer_name"] = df["customer_name"].str.strip().str.title()
df["city"] = df["city"].str.strip().str.title()

df["city"] = df["city"].replace({
    "Jed": "Jeddah",
    "Ryd": "Riyadh"
})

print(df["city"].value_counts())
```

Q4:
The dataset started with 10 distinct cities and ended with 3 distinct cities

```python
duplicates = df.duplicated().sum()

df = df.drop_duplicates()

df = df.dropna(subset=["total"])

df = df[df["total"] >= 0]

df = df[df["rating"] <= 5]

print("Duplicate rows removed:", duplicates)
```
